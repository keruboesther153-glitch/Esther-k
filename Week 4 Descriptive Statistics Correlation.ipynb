## 1. Import the required libraries
Pandas is used to work with the data, NumPy for numerical operations, and Matplotlib for the plot.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda value: f'{value:.3f}')
```

## 2. Load and inspect the dataset
The dataset has no header row in the raw CSV, so the column names are supplied below. If the online CSV is unavailable, upload `diabetes.csv` to the notebook environment and use the fallback line shown in the code.

```python
DATA_URL = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv'
columns = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
    'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome'
]

try:
    df = pd.read_csv(DATA_URL, header=None, names=columns)
except Exception as error:
    print('Could not load the online dataset:', error)
    # If you uploaded the CSV, uncomment the next line:
    # df = pd.read_csv('diabetes.csv')
    raise

print('Dataset shape:', df.shape)
display(df.head())
display(df.dtypes.to_frame('Data type'))
```

## 3. Select numerical variables
Four numerical variables relevant to the diabetes project are selected: **Glucose, BMI, BloodPressure, and Age**. The target `Outcome` is not included in the descriptive-statistics calculations below.

```python
selected_variables = ['Glucose', 'BMI', 'BloodPressure', 'Age']
data = df[selected_variables].copy()

# In this dataset, zero values for these measurements are treated as missing
# because they generally do not represent plausible measured values.
measurements = ['Glucose', 'BMI', 'BloodPressure']
data[measurements] = data[measurements].replace(0, np.nan)

print('Selected variables:', selected_variables)
print('Missing values in selected variables:')
display(data.isna().sum().to_frame('Missing values'))
display(data.head())
```

## 4. Calculate descriptive statistics
For each variable, calculate the mean, median, range, variance, and standard deviation. Missing values are excluded from these calculations. Pandas uses sample variance and sample standard deviation by default (`ddof=1`).

```python
statistics = pd.DataFrame(index=selected_variables)
statistics['Mean'] = data.mean()
statistics['Median'] = data.median()
statistics['Minimum'] = data.min()
statistics['Maximum'] = data.max()
statistics['Range'] = statistics['Maximum'] - statistics['Minimum']
statistics['Variance'] = data.var()
statistics['Standard deviation'] = data.std()

display(statistics)
```

### What do the statistics tell us?
- **Mean:** the average value of a variable.
- **Median:** the middle value after sorting the observations; it is less affected by extreme values than the mean.
- **Range:** the difference between the largest and smallest observed values; a larger range means the data cover a wider span.
- **Variance and standard deviation:** describe how spread out the observations are. A larger standard deviation indicates greater variation in the variable's original units.

Use the table above to identify which selected variable has the largest range and standard deviation. These statistics describe this dataset; they do not by themselves explain why the measurements differ.

## 5. Calculate covariance and correlation
Glucose and BMI are selected to examine whether their values tend to vary together. The calculations use rows where both values are present.

```python
pair = data[['Glucose', 'BMI']].dropna()
covariance = pair['Glucose'].cov(pair['BMI'])
correlation = pair['Glucose'].corr(pair['BMI'])

print(f'Number of complete Glucose–BMI pairs: {len(pair)}')
print(f'Covariance between Glucose and BMI: {covariance:.3f}')
print(f'Correlation between Glucose and BMI: {correlation:.3f}')
```

### How should correlation be interpreted?
Correlation ranges from **-1 to +1**. A positive value indicates that higher values of one variable tend to occur with higher values of the other; a negative value indicates that higher values of one tend to occur with lower values of the other. A value close to zero indicates a weak linear relationship. The strength is judged by how close the value is to -1 or +1.

**Covariance** indicates the direction in which two variables vary together, but its size depends on the variables' units. Correlation is easier to compare because it is standardized.

**Correlation does not prove causation.** Even if Glucose and BMI are correlated, this does not establish that one causes the other. Other factors, the way the data were collected, or chance may help explain the relationship.

## 6. Plot the relationship between the two variables
The scatter plot shows each available Glucose–BMI pair. Each point represents one observation.

```python
plt.figure(figsize=(8, 5))
plt.scatter(pair['Glucose'], pair['BMI'], alpha=0.6)
plt.xlabel('Glucose')
plt.ylabel('BMI')
plt.title(f'Relationship between Glucose and BMI (r = {correlation:.3f})')
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()
```

### What does the plot show?
Look at the direction and spread of the points. An upward pattern suggests a positive relationship, a downward pattern suggests a negative relationship, and a widely scattered cloud suggests a weak linear relationship. Compare your visual observation with the calculated correlation coefficient above. Do not claim a strong relationship unless the plot and coefficient support that conclusion.
